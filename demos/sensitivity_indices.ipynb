# Sobol' Sensitivity Indices

Computes first-order and total Sobol' sensitivity indices for the Ishigami function using the pick-freeze estimator with QMC sampling.

```julia
using QMC
import QMC: Uniform
```

## The Ishigami Function

f(x) = sin(x₁) + 7 sin²(x₂) + 0.1 x₃⁴ sin(x₁), where x ∈ [-π, π]³.

Known analytical indices: S₁ ≈ 0.314, S₂ ≈ 0.442, S₃ ≈ 0.000.

```julia
# Pick-freeze requires 2d dimensions
dd = IIDStdUniform(6; seed=7)
tm = Uniform(dd)
base = Ishigami(tm)
si = SensitivityIndices(base)
println("SensitivityIndices integrand wrapping Ishigami(3D)")
```

## Compute Indices

```julia
# Integrate with CubMCCLTVec
sc = CubMCCLTVec(si; abs_tol=0.02)
result = integrate(sc)
println("Solution: $(result.solution)")
println("Converged: $(result.data[:converged])")
println("Total samples: $(get(result.data, :n_total, get(result.data, :n, 0)))")
```
