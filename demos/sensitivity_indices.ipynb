# Sobol' Sensitivity Indices

Computes first-order and total Sobol' sensitivity indices for the Ishigami function using the pick-freeze estimator with QMC sampling.

```julia
using QMC
import QMC: Uniform
```

## The Ishigami Function

$$f(\mathbf{x}) = \sin(x_1) + 7\sin^2(x_2) + 0.1\,x_3^4\sin(x_1), \quad \mathbf{x} \in [-\pi, \pi]^3$$

Known analytical indices: $S_1 \approx 0.314$, $S_2 \approx 0.442$, $S_3 \approx 0.000$ (first-order); $T_1 \approx 0.558$, $T_2 \approx 0.442$, $T_3 \approx 0.244$ (total).

```julia
# Pick-freeze uses 2d independent samples; Ishigami lives on [-π, π]³
d = 3
dd = IIDStdUniform(2d; seed=7)
tm = Uniform(dd; lower_bound=-π, upper_bound=π)
base = Ishigami(tm)
si = SensitivityIndices(base)
println(si)
```

## Compute Indices

```julia
using Printf

# Generate n pick-freeze samples and compute indices directly.
# n = 4096 gives standard errors well below abs_tol = 0.02.
n = 4096
x = gen_samples(dd, n)
xt = transform(tm, x)
closed, total = compute_sensitivity_indices(si, xt)
ref = ishigami_exact()

println("Sobol' sensitivity indices — Ishigami (n = $n, abs_tol ≈ 0.02)")
println()
@printf("  %-5s  %8s  %8s  │  %8s  %8s\n", "dim", "S (est)", "S (exact)", "T (est)", "T (exact)")
println("  " * repeat("─", 46))
for i in 1:d
    @printf("  x_%d    %8.2f  %8.2f  │  %8.2f  %8.2f\n",
        i, closed[i], ref.closed[i], total[i], ref.total[i])
end
```
