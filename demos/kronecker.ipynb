# Kronecker Low-Discrepancy Sequence

The Kronecker sequence uses irrational number multiples (√primes) to generate well-distributed points in the unit cube.

```julia
using QMC
using Statistics
```

## Generate Points

```julia
# 2D Kronecker sequence
kr = Kronecker(2; seed=42)
x = gen_samples(kr, 1000)

println("Kronecker sequence: $(size(x, 1)) points in $(size(x, 2))D")
println("Mean: $(round.(mean(x, dims=1), digits=4))")
println("Std:  $(round.(std(x, dims=1), digits=4))")
```

## Integration Comparison

Compare Kronecker vs IID for integrating f(x) = ∏ sin(πxᵢ) (true value = 0).

```julia
kr_integrand(x) = prod(sin.(π .* x))

# Kronecker estimate
kr_est = mean(kr_integrand(x[i, :]) for i in 1:size(x, 1))
println("Kronecker estimate: $(round(kr_est, digits=6))")

# IID MC estimate
dd_iid = IIDStdUniform(2; seed=42)
x_iid = gen_samples(dd_iid, 1000)
iid_est = mean(kr_integrand(x_iid[i, :]) for i in 1:size(x_iid, 1))
println("IID MC estimate:    $(round(iid_est, digits=6))")
println("True value:         0.0")
```
